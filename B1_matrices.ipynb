## Fundamental Subspaces of a Matrix

A matrix $A \in \mathbb{R}^{n \times p}$ represents a linear function $f$ from $\mathbb{R}^p$ to $\mathbb{R}^n$. The image set of $f$ (and $A$), i.e., the set of all function outputs

$$\mathrm{im}(A)=\mathrm{im}(f)=\{Ax \!: \, x \in \mathbb{R}^p\}$$

is also called the **column space** $A$. The word "space" is justified by the fact that $\mathrm{im}(A)$ is a complete vector space. Indeed, $\mathbf{0} = A\mathbf{0} \in \mathrm{im}(A)$, $-y=A(-x) \in \mathrm{im}(A)$ for all $y \in \mathrm{im}(A)$, and the linearity of the function implies that if $y_1=Ax_1$ and $y_2=Ax_2$ are an element of $\mathrm{im}(A)$, then also $$\alpha y_1 + \beta y_2 = \alpha Ax_1 + \beta Ax_2 = A(\alpha x_1 + \beta x_2) \enspace .$$

Further, the set of all function inputs that is mapped to the zero output vector $\mathbf{0} \in $, kernel or **null space** of $A$

$$\mathrm{null}(A)=\{x \in \mathbb{R}^p \! : \, Ax=\mathbf{0}\}$$
