# Singular Value Decomposition

Any matrix $X \in \mathbb{R}^{n \times p}$ with rank $r$ can be decomposed into a product

$$
X = USV^\top
$$

with

1. a diagonal matrix $S = \mathrm{diag}(s_1, \dots, s_r) \in \mathbb{R}^{r \times r}$ with positive numbers $s_1, \dots, s_r > 0$ called **singular values**
2. semi-orthogonal column matrices $U \in \mathbb{R}^{n \times r}$ and $V \in \mathbb{R}^{p \times r}$, the columns of which are called the left and right **singular vectors**, respectively.


```python
import numpy as np

x = np.array([[1, 0, 2],
              [-1, 2, 3],
              [-3, -1, 2],
              [2, 2, -1]])

np.linalg.svd(x[:, [0]], full_matrices=False)
```

```python
np.linalg.svd(x[:, [0, 1]], full_matrices=False)
```

```python
np.linalg.svd(x[:, [0, 1, 2]], full_matrices=False)
```
