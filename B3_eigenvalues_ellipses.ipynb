## Hyperspheres

Hyperspheres are generalisations of circles in an arbitrary dimension $n$ with the special cases of circles ($n=2$) and spheres ($n=3$). The definition of a hypersphere of radius $s>0$ centred at the origin is the set of points $x$ with distance $s$ from the origin or in other words all vectors $x$ of length or Euclidean norm $\|x\|=s$:

$$
H = \left\{\mathbf x \in \mathbb{R}^n \, : \, \|\mathbf x\| = \sqrt{x^2_1 + \dots + x_n^2} = s \right\}
$$

Alternatively, we can think of the same circle as a *unit circle*, i.e., of radius $1$, with respect to a scaled coordinate system, where coordinates represent multiples of $s$:

$$
\begin{aligned}
H &= \left\{\mathbf x \in \mathbb{R}^n \, : \, \|\mathbf x\|^2 = x^2_1 + \dots + x_n^2 = s^2 \right\}\\
&=\left\{\mathbf x \in \mathbb{R}^n \, : \, \|s^{-1}\mathbf x\|^2 = \left(\frac{x_1}{s}\right)^2 + \dots + \left(\frac{x_n}{s}\right)^2 = 1 \right\}
\end{aligned}
$$

## Standard Form Ellipsoids

The last form is easily generalised to **ellipsoids**. These are generalisations of ellipses from two dimensions ($n=2$) to arbitrary dimension n. They arise from replacing the uniform scaling factor $s$ in the circle definition with different factors $s_1, \dots, s_n$ per coordinate axis:

$$
\begin{aligned}
E &= \left\{\mathbf x \in \mathbb{R}^n \,:\, \left(\frac{x_1}{s_1}\right)^2 + \dots + \left(\frac{x_n}{s_n}\right)^2 = 1\right\}\\
&= \left\{\mathbf x \in \mathbb{R}^n \,:\, \|\mathbf S^{-1}\mathbf x\| = 1\right\}\\
\end{aligned}
$$

where $\mathbf S$ is the diagonal matrix with entries $s_1, \dots, s_n$, i.e.,

$$\mathbf S=\mathrm{diag}(s_1, \dots, s_n)=
\begin{bmatrix}
s_1 & 0 & \dots & 0\\
0 & s_2 & \dots & 0\\
\vdots & & \ddots & \vdots \\
0 & 0 & \dots & s_n
\end{bmatrix}$$

and the inverse matrix $\mathbf S^{-1}$ is therefore the diagonal matrix with entries $s_1^{-1}, \dots s_n^{-1}$.

## General Centred Ellipsoids

$$
\begin{aligned}
E &= \left\{\mathbf{x} \in \mathbb{R}^n \, : \, \left( \frac{\mathbf{v}_1^\top\mathbf{x}}{s_1}\right)^2 + \cdots + \left( \frac{\mathbf{v}_n^\top\mathbf{x}}{s_n}\right)^2 = 1\right \}\\
  &= \left\{\mathbf{x} \in \mathbb{R}^n \, : \, \|\mathbf{S}^{-1}\mathbf{V}^\top\mathbf{x}\|^2=1 \right\}
\end{aligned}
$$

## Ellipsoid Norm

Let $A=\mathbf V\mathbf S^{-2}\mathbf V^\top$

$$
\begin{aligned}
\|\mathbf S^{-1} \mathbf V^\top \mathbf x\| &= \sqrt{\mathbf{x}^\top\mathbf V \mathbf S^{-2} \mathbf{V}^\top\mathbf x}\\
&= \sqrt{\mathbf x^\top\mathbf A\mathbf x}\\
&= \|\mathbf x\|_{\mathbf A}
\end{aligned}
$$

```python
xx1, xx2 = np.meshgrid(np.linspace(-4, 4, 100), np.linspace(-4, 4))

def ellipsoid_norm(x, s, v):
    t = np.diag(1/s) @ v.T
    x_t = t @ x.T
    return np.linalg.norm(x_t, axis=0)

ellipsoid_norm(np.array([[1, 1], [2, 2], [3, 3]]), np.array([a1, a2]), v)
```
